```python
import numpy as np
```

```python

def func1(theta):
    return np.sin(theta)+np.cos(theta)

def func2(x, y):
    return 5*(x**2) + (y**3)

g=32.2
def func3(t, s):
    return s-(0.5*g*(t**2))

def func4(r):
    return np.pi*(r**2)

def func5(t):
    position = 2*t**3-15*t**2+36*t+10
    velocity = 6*t**2 - 30*t + 36
    acceleration = 12*t - 30
    return (position, velocity, acceleration)

print(func1(90))
print(func2(3, 2))
print(func3(5, 60))
print(func4(3))
for i in range (0,6):
    print(func5(i))
```
